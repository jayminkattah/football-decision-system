## Limitations to mention in the portfolio write-up

- The strategy is evaluated using walk-forward predictions, not random shuffled validation.
- Betting markets are strong baselines, so beating them consistently is hard.
- Flat staking avoids capital allocation overengineering.
- No threshold tuning is done on the full test set.
- Results should be interpreted as a decision-system demonstration, not a money printer. Money printers are usually just Excel sheets with confidence issues.
