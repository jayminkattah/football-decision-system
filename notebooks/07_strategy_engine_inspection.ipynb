# Day 7 — Strategy Engine Inspection

Purpose: inspect the long-format opportunity table created by `scripts/build_strategy_opportunities.py`.

No transformations should live here. This notebook only reads outputs and checks sanity.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

ROOT = Path('..').resolve()
OPPORTUNITIES_PATH = ROOT / 'data' / 'processed' / 'strategy_opportunities.parquet'
REPORT_PATH = ROOT / 'outputs' / 'evaluation' / 'strategy_report.csv'
MIN_EDGE = 0.02
```

```python
opportunities = pd.read_parquet(OPPORTUNITIES_PATH)
report = pd.read_csv(REPORT_PATH)

opportunities.shape, report.shape
```

```python
opportunities['outcome'].value_counts().sort_index()
```

```python
opportunities.groupby('outcome')['selected'].sum().sort_index()
```

```python
expected_selected = opportunities['edge'].ge(MIN_EDGE).fillna(False)
assert opportunities['selected'].equals(expected_selected)
print('Selected flag matches edge >= min_edge rule.')
```

```python
assert opportunities.groupby('match_id').size().eq(3).all()
print('Every match has exactly three outcome rows.')
```

```python
selected = opportunities.loc[opportunities['selected']].copy()
selected.sort_values('edge', ascending=False).head(20)
```

```python
selected_counts = opportunities.groupby('outcome')['selected'].sum().reindex(['H', 'D', 'A'])
selected_counts.plot(kind='bar')
plt.title('Selected Opportunities by Outcome')
plt.xlabel('Outcome')
plt.ylabel('Selected count')
plt.show()
```
