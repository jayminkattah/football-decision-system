# Day 9 — Walk-forward Backtest Inspection

This notebook inspects saved backtest outputs only.

No production transformations live here. No threshold tuning. No backtest archaeology where we pretend the future was unknowable after staring directly at it.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

by_season_path = Path('../outputs/backtests/backtest_by_season.csv')
cumulative_path = Path('../outputs/backtests/backtest_cumulative.csv')
report_path = Path('../outputs/evaluation/backtest_report.csv')

by_season = pd.read_csv(by_season_path)
cumulative = pd.read_csv(cumulative_path, parse_dates=['date'])
report = pd.read_csv(report_path)
```

## Overall report

```python
report
```

## Season-level performance

```python
by_season
```

## Cumulative profit over walk-forward bets

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(cumulative['date'], cumulative['cumulative_profit'])
ax.axhline(0, linestyle='--', linewidth=1)
ax.set_title('Cumulative Profit — Walk-forward Backtest')
ax.set_xlabel('Date')
ax.set_ylabel('Cumulative Profit')
plt.xticks(rotation=45)
plt.tight_layout()
```

## Drawdown over time

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(cumulative['date'], cumulative['drawdown'])
ax.set_title('Backtest Drawdown')
ax.set_xlabel('Date')
ax.set_ylabel('Drawdown')
plt.xticks(rotation=45)
plt.tight_layout()
```

## Inspection notes

- These results aggregate simulated bets already produced from walk-forward predictions.
- The notebook does not change strategy thresholds.
- Season-level instability matters. One lucky season is not a strategy; it is a spreadsheet wearing perfume.
